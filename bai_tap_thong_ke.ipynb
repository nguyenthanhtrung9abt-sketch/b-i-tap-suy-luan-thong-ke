## 3a. Ước lượng kỳ vọng tổng thể bằng khoảng tin cậy 95%
**Biến lựa chọn:** Cân nặng của chim cánh cụt (`body_mass_g`).

Vì chưa biết phương sai tổng thể, ta sử dụng phân phối Student (T-distribution). Khoảng tin cậy 95% cho kỳ vọng $\mu$ được tính bằng công thức:
$$\mu \in \left( \overline{x} - t_{\alpha/2, n-1} \cdot \frac{s}{\sqrt{n}} ; \overline{x} + t_{\alpha/2, n-1} \cdot \frac{s}{\sqrt{n}} \right)$$

Trong đó:
* $\overline{x}$: Trung bình mẫu
* $s$: Độ lệch chuẩn mẫu
* $n$: Kích thước mẫu
* $t_{\alpha/2, n-1}$: Giá trị tới hạn T với độ tin cậy 95% ($\alpha = 0.05$)

```python
# CHỈ ĐỊNH RÕ CỘT VÀ XỬ LÝ DỮ LIỆU TRỐNG
data_can_nang = df["body_mass_g"].dropna()

# 1. Tính toán các đại lượng đặc trưng
n = len(data_can_nang)               # Cỡ mẫu (số lượng chim)
mean_val = data_can_nang.mean()      # Trung bình mẫu (x ngang)
std_val = data_can_nang.std(ddof=1)  # Độ lệch chuẩn mẫu (s)

# 2. Tính khoảng tin cậy 95%
confidence = 0.95
alpha = 1 - confidence

# Tìm giá trị t tới hạn (t-critical) từ bảng phân phối Student
t_crit = stats.t.ppf(1 - alpha/2, df=n-1)

# Tính sai số biên (Margin of Error)
margin_of_error = t_crit * (std_val / np.sqrt(n))

# 3. Giới hạn dưới và giới hạn trên của khoảng tin cậy
lower_bound = mean_val - margin_of_error
upper_bound = mean_val + margin_of_error

print("--- KẾT QUẢ CÂU 3a ---")
print(f"Cỡ mẫu (n): {n}")
print(f"Trung bình mẫu (x̄): {mean_val:.2f} (gram)")
print(f"Độ lệch chuẩn (s): {std_val:.2f} (gram)")
print(f"Khoảng tin cậy 95% cho kỳ vọng: ({lower_bound:.2f}, {upper_bound:.2f})")
```

### Diễn giải kết quả Câu 3a
*   **Điều kiện áp dụng:** Kích thước mẫu lớn ($n = 333 > 30$)[cite: 6]. Theo Định lý Giới hạn Trung tâm (Central Limit Theorem), phân phối của trung bình mẫu xấp xỉ phân phối chuẩn[cite: 6]. Vì phương sai của tổng thể chưa biết, ta sử dụng phân phối Student (t-distribution) để xây dựng khoảng tin cậy[cite: 6]. Dữ liệu được giả định là lấy mẫu ngẫu nhiên độc lập[cite: 6].
*   **Kết quả tính toán:** Từ dữ liệu thực tế, ta tính được trung bình mẫu $\bar{x} = 4207.06$ gram, độ lệch chuẩn $s = 805.22$ gram[cite: 6]. Khoảng tin cậy 95% tính được là: (4120.26, 4293.86)[cite: 6].
*   **Kết luận:** Ta có tự tin 95% rằng, khối lượng trung bình thực sự của toàn bộ quần thể chim cánh cụt (tương tự như các loài trong mẫu) nằm trong khoảng từ 4120.26 gram đến 4293.86 gram[cite: 6].

## 3b. Ước lượng phương sai tổng thể bằng khoảng tin cậy 95%
**Biến lựa chọn:** Cân nặng của chim cánh cụt (`body_mass_g`).

Để ước lượng phương sai tổng thể ($\sigma^2$), ta sử dụng phân phối Chi-bình phương ($\chi^2$) với bậc tự do $n-1$. Điều kiện áp dụng là biến tổng thể phải có phân phối xấp xỉ chuẩn.

Công thức khoảng tin cậy 95% cho phương sai $\sigma^2$:
$$\sigma^2 \in \left( \frac{(n-1)s^2}{\chi^2_{1-\alpha/2, n-1}} ; \frac{(n-1)s^2}{\chi^2_{\alpha/2, n-1}} \right)$$

Trong đó:
* $s^2$: Phương sai mẫu
* $n$: Kích thước mẫu
* $\chi^2_{1-\alpha/2, n-1}$ và $\chi^2_{\alpha/2, n-1}$: Các giá trị tới hạn tra từ phân phối Chi-bình phương. Do phân phối này không đối xứng, ta phải tra hai giá trị riêng biệt cho hai bên đuôi (mỗi đuôi chiếm diện tích $\alpha/2 = 0.025$).

```python
# 1. Tính phương sai mẫu (s^2) 
# Dùng lại biến data_can_nang đã làm sạch từ Câu 3a
var_val = data_can_nang.var(ddof=1)  

# 2. Tìm giá trị tới hạn Chi-bình phương
# Sử dụng alpha = 0.05 và n đã có từ Ô số 4
chi2_lower = stats.chi2.ppf(1 - alpha/2, df=n-1) 
chi2_upper = stats.chi2.ppf(alpha/2, df=n-1)      

# 3. Tính Cận dưới và Cận trên cho phương sai
var_lower_bound = ((n - 1) * var_val) / chi2_lower
var_upper_bound = ((n - 1) * var_val) / chi2_upper

print("--- KẾT QUẢ CÂU 3b ---")
print(f"Phương sai mẫu (s^2): {var_val:.2f}")
print(f"Giá trị Chi-square dưới: {chi2_lower:.2f}")
print(f"Giá trị Chi-square trên: {chi2_upper:.2f}")
print(f"=> Khoảng tin cậy 95% cho phương sai tổng thể: ({var_lower_bound:.2f}, {var_upper_bound:.2f})")
```

### Diễn giải kết quả Câu 3b
*   **Điều kiện áp dụng:** Để ước lượng phương sai tổng thể bằng khoảng tin cậy, ta giả định phân phối của tổng thể xấp xỉ phân phối chuẩn (Normal distribution). Đại lượng thống kê được sử dụng tuân theo phân phối Chi-bình phương (Chi-square distribution) với $n - 1$ bậc tự do.
*   **Kết quả tính toán:** Từ dữ liệu thu thập, ta tính được phương sai mẫu $s^2 = 648372.49$. Khoảng tin cậy 95% tính được là: (560030.16, 759519.99).
*   **Kết luận:** Ta có mức độ tin cậy 95% rằng, phương sai khối lượng của toàn bộ quần thể chim cánh cụt tại Nam Cực rơi vào khoảng từ 560030.16 đến 759519.99. Con số này phản ánh mức độ biến động về cân nặng của chúng trong môi trường tự nhiên.

## 3c. Ước lượng tỷ lệ tổng thể cho biến định tính (nhị phân) bằng khoảng tin cậy 95%
**Biến lựa chọn:** Giới tính của chim cánh cụt (`sex`), phân loại thành 2 nhóm MALE và FEMALE.
**Mục tiêu:** Ước lượng tỷ lệ chim mái (FEMALE) trong tổng thể.

*   **Điều kiện áp dụng:** Kích thước mẫu đủ lớn để có thể xấp xỉ phân phối nhị thức bằng phân phối chuẩn. Điều kiện là số lượng thành công ($n \cdot \hat{p}$) và thất bại ($n \cdot (1-\hat{p})$) đều phải $\ge 5$. Ở đây mẫu có kích thước rất lớn ($n=333$), điều kiện này hoàn toàn thỏa mãn. Do đó, ta sử dụng phân phối Chuẩn hóa (Z-distribution).
*   **Công thức:**
$$p \in \left( \hat{p} - Z_{\alpha/2} \cdot \sqrt{\frac{\hat{p}(1-\hat{p})}{n}} ; \hat{p} + Z_{\alpha/2} \cdot \sqrt{\frac{\hat{p}(1-\hat{p})}{n}} \right)$$

Trong đó:
* $\hat{p}$: Tỷ lệ mẫu (số lượng chim cái chia cho tổng số chim)
* $n$: Kích thước mẫu
* $Z_{\alpha/2}$: Giá trị tới hạn từ phân phối Chuẩn hóa (Standard Normal) với độ tin cậy 95%.

```python
# CHUẨN BỊ DỮ LIỆU
# Lấy cột giới tính, bỏ các giá trị trống (NaN) và bỏ giá trị lỗi "." đã phát hiện ở phần Mô tả
data_sex = df['sex'].dropna()
data_sex = data_sex[data_sex != "."]

# 1. Tính các đại lượng cơ bản
n_sex = len(data_sex)                            # Tổng số chim đã xác định giới tính
so_chim_mai = (data_sex == 'FEMALE').sum()       # Đếm số lượng chim cái (FEMALE)
p_hat = so_chim_mai / n_sex                      # Tỷ lệ chim cái trong mẫu

# 2. Tính giá trị tới hạn Z từ phân phối chuẩn hóa
# Vì dùng phân phối Chuẩn (Normal) nên gọi stats.norm thay vì stats.t
alpha = 0.05
z_critical = stats.norm.ppf(1 - alpha/2)

# 3. Tính sai số biên (Margin of Error) cho tỷ lệ
# Công thức: Z * căn bậc hai của [p_hat * (1 - p_hat) / n]
margin_of_error_prop = z_critical * np.sqrt((p_hat * (1 - p_hat)) / n_sex)

# 4. Giới hạn dưới và trên của khoảng tin cậy
lower_bound_prop = p_hat - margin_of_error_prop
upper_bound_prop = p_hat + margin_of_error_prop

# IN KẾT QUẢ
print("--- KẾT QUẢ CÂU 3c ---")
print(f"Tổng số chim (n): {n_sex}")
print(f"Số lượng chim cái: {so_chim_mai}")
print(f"Tỷ lệ mẫu (p_hat): {p_hat:.4f} (hay {p_hat*100:.2f}%)")
print(f"Giá trị tới hạn Z: {z_critical:.2f}")
print(f"Khoảng tin cậy 95% cho tỷ lệ: ({lower_bound_prop:.4f}, {upper_bound_prop:.4f})")
print(f"                                (hay {lower_bound_prop*100:.2f}% đến {upper_bound_prop*100:.2f}%)")
```

### Diễn giải kết quả Câu 3c
*   **Kết quả tính toán:** Từ mẫu gồm 333 cá thể hợp lệ, ta đếm được 165 chim mái, tương đương tỷ lệ mẫu là $\hat{p} = 49.55\%$.
*   Khoảng tin cậy 95% cho tỷ lệ chim mái tổng thể tính được là $(0.4418, 0.5492)$, tức là từ $44.18\%$ đến $54.92\%$.
*   **Kết luận:** Ta có mức độ tin cậy 95% rằng tỷ lệ chim mái thực sự trong toàn bộ quần thể chim cánh cụt ở Nam Cực sẽ rơi vào khoảng từ $44.18\%$ đến $54.92\%$. Con số này cho thấy tỷ lệ đực/cái trong tự nhiên khá cân bằng (xấp xỉ mức 50%).

## 3d. So sánh kỳ vọng giữa hai nhóm bằng khoảng tin cậy cho hiệu kỳ vọng $\mu_1 - \mu_2$
**Biến phân nhóm (định tính):** Giới tính `sex` (MALE và FEMALE).
**Biến so sánh (định lượng):** Cân nặng `body_mass_g`.
**Mục tiêu:** Xây dựng khoảng tin cậy 95% cho sự chênh lệch cân nặng trung bình giữa chim đực ($\mu_1$) và chim mái ($\mu_2$).

*   **Điều kiện áp dụng:** Hai mẫu (đực và mái) là hoàn toàn độc lập. Kích thước cả hai mẫu đều lớn ($n > 30$). Vì không thể giả định phương sai cân nặng của hai giới tính là bằng nhau, ta áp dụng phương pháp Welch (Welch's t-interval) để tính toán chính xác nhất.
*   **Công thức:**
$$(\mu_1 - \mu_2) \in \left( (\bar{x}_1 - \bar{x}_2) - t_{\alpha/2, df} \cdot SE ; (\bar{x}_1 - \bar{x}_2) + t_{\alpha/2, df} \cdot SE \right)$$

Trong đó:
* $\bar{x}_1, \bar{x}_2$: Trung bình mẫu của 2 nhóm.
* $SE$ (Sai số chuẩn): $SE = \sqrt{\frac{s_1^2}{n_1} + \frac{s_2^2}{n_2}}$
* Bậc tự do $df$ được tính theo phương trình Welch-Satterthwaite.

```python
# CHUẨN BỊ DỮ LIỆU
# Lọc bỏ các dòng thiếu giới tính hoặc thiếu cân nặng, bỏ giá trị lỗi "."
df_clean = df.dropna(subset=['sex', 'body_mass_g'])
df_clean = df_clean[df_clean['sex'] != "."]

# Tách thành 2 nhóm
chim_duc = df_clean[df_clean['sex'] == 'MALE']['body_mass_g']
chim_mai = df_clean[df_clean['sex'] == 'FEMALE']['body_mass_g']

# 1. Thống kê mô tả cho từng nhóm
n1, mean1, s1 = len(chim_duc), chim_duc.mean(), chim_duc.std(ddof=1)
n2, mean2, s2 = len(chim_mai), chim_mai.mean(), chim_mai.std(ddof=1)

# 2. Tính hiệu hai trung bình mẫu
diff_mean = mean1 - mean2

# 3. Tính sai số chuẩn (SE) và bậc tự do (df) theo công thức Welch
se = np.sqrt((s1**2 / n1) + (s2**2 / n2))
df_welch = (se**4) / (((s1**2 / n1)**2 / (n1 - 1)) + ((s2**2 / n2)**2 / (n2 - 1)))

# 4. Tìm giá trị tới hạn T và tính sai số biên
alpha = 0.05
t_crit = stats.t.ppf(1 - alpha/2, df=df_welch)
margin_of_error_diff = t_crit * se

# 5. Giới hạn khoảng tin cậy
lower_bound_diff = diff_mean - margin_of_error_diff
upper_bound_diff = diff_mean + margin_of_error_diff

# IN KẾT QUẢ
print("--- KẾT QUẢ CÂU 3d ---")
print(f"Nhóm Đực: n1 = {n1}, trung bình = {mean1:.2f}g, độ lệch chuẩn = {s1:.2f}g")
print(f"Nhóm Mái: n2 = {n2}, trung bình = {mean2:.2f}g, độ lệch chuẩn = {s2:.2f}g")
print(f"Hiệu trung bình mẫu (Đực - Mái): {diff_mean:.2f}g")
print(f"Bậc tự do (Welch df): {df_welch:.2f}")
print(f"=> Khoảng tin cậy 95% cho chênh lệch (μ1 - μ2): ({lower_bound_diff:.2f}, {upper_bound_diff:.2f})")
```

*   **Kết quả tính toán:** Trung bình mẫu cho thấy chim đực nặng hơn chim mái $683.41$ gram. Khoảng tin cậy 95% cho sự chênh lệch (Hiệu kỳ vọng $\mu_1 - \mu_2$) là từ $526.25$ gram đến $840.58$ gram.
*   **Nhận xét cốt lõi:** Ta thấy toàn bộ khoảng tin cậy $(526.25, 840.58)$ đều mang **giá trị dương** (không chứa số 0). Điều này có ý nghĩa thống kê rất quan trọng.
*   **Kết luận:** Với độ tin cậy 95%, ta có đủ cơ sở để kết luận rằng: **Chim cánh cụt đực có cân nặng trung bình thực sự trong tự nhiên lớn hơn chim cánh cụt mái**. Mức chênh lệch khối lượng rơi vào khoảng từ xấp xỉ 526 gram đến 840 gram.
